RANK


```python
import pandas as pd
from pandas import Series

x = [90, 54, 55, 60, 75, 75,75,90,95]

y = pd.Series(x)

print(y)

print()

print(y.rank(ascending=False))

print()

q=y.rank(ascending=True)

print()

print(q.rank(method="max"))

print(q.rank(method="min"))
print(q.rank(method="first"))

print(q.rank(method="dense"))


```

```python
s=[99,95,85,75,75,65]
d=pd.Series(s)
z=d.rank(ascending=False)
print(z)
print(z.rank(method="average"))

print(z.rank(method="max"))
print(z.rank(method="min"))
print(z.rank(method="first"))
print(z.rank(method="dense"))



```

```python
import pandas as pd
name=["ram","sham","viraj","rohit","virat","suraj","swayam","mayur","hari","sager"]
gender=["male","male","female","male","female","male","male","male","female","female"]
course=["python","java","python","java","python","java","python","java","python","java"]
fees=[4500,3500,2500,1500,7500,8000,4000,6000,5000,9500]
data={"student_name":name,
      "gender":gender,
      "course":course,
      "fees":fees}
x=pd.DataFrame(data)
print(x)
print()
print()

print(x.groupby("course").sum())
print()
print()
print(x.groupby("course")["fees"].count())
print()
print()
print(x.groupby("course")["fees"].max())
print()
print()
print(x.groupby("course")["fees"].min())
print()
print()
print(x.groupby("course")["fees"].agg(["sum","max","min","count","mean"]))

print()
print()
print(x.groupby("course")["fees"].agg(["sum","count","min","max"]).loc["python"])
```

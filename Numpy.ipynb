```python
#asarray()--we want convert from one data type to another data type
#new_var=np.asarray(array,dtype=datatype)

x=[1,2,3,4]
x1=np.asarray(x,dtype=float)#float datatypr convert
print(x1)
```

```python
#arange()
#it is an inbuilt method which is used to the sequence of int
#between the specified limit

#syntax:--->np.arange(si,ei+1,stepvalue)


x=np.arange(1,11,1)
print(x)
```

```python
#linspace
#It is an inbuilt method which is used to create the equally
#space values between the given limits


#syntax:--> np.linspace(si,ei,num,retstep=true)

x=np.linspace(1,10,5,retstep=True)
print(x)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris=pd.read_csv('/content/iris.csv')
iris.head(100)
```

```python
iris['Species']=iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='summer',alpha=1)

plt.xlabel('SepalLenghCM')
plt.ylabel('PetalLengthCM')

plt.colorbar()
```

```python
plt.figure(figsize=(15,7))


plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='summer',alpha=1)

plt.xlabel('SepalLenghCM')
plt.ylabel('PetalLengthCM')

plt.colorbar()
```
